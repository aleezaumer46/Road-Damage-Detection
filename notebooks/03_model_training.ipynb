# **Set Dataset Path**

```python
DATASET_PATH = "/kaggle/input/datasets/aliabdelmenam/rdd-2022/RDD_SPLIT"
```

# **Check Classes**

```python
import os

labels_dir = f"{DATASET_PATH}/train/labels"

classes = set()

for file in os.listdir(labels_dir):
    if file.endswith(".txt"):
        with open(os.path.join(labels_dir, file), "r") as f:
            for line in f:
                if line.strip():
                    classes.add(int(line.split()[0]))

print("Classes:", sorted(classes))
print("Number of Classes:", len(classes))
```

# **Create data.yaml**

```python
%%writefile /kaggle/working/data.yaml

train: /kaggle/input/datasets/aliabdelmenam/rdd-2022/RDD_SPLIT/train/images
val: /kaggle/input/datasets/aliabdelmenam/rdd-2022/RDD_SPLIT/val/images
test: /kaggle/input/datasets/aliabdelmenam/rdd-2022/RDD_SPLIT/test/images

nc: 5

names:
  0: D00
  1: D10
  2: D20
  3: D40
  4: D43
```

# **Set YAML Path**

```python
DATA_YAML = "/kaggle/working/data.yaml"
```

# **Load YOLO Model**

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")
```

# **Train Model**

```python
results = model.train(
    data=DATA_YAML,
    epochs=50,
    imgsz=640,
    batch=16,
    workers=4,
    patience=20,
    device=0,
    project="runs",
    name="road_damage_detection",
    pretrained=True,
    save=True,
    verbose=True
)
```
